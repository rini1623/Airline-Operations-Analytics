```python
delayed_flights = df[df["ARR_DELAY"] > 15]

print("Total delayed flights:", len(delayed_flights))
```

```python
delay_percentage = (len(delayed_flights) / len(df)) * 100

print("Delayed flight percentage:", round(delay_percentage, 2), "%")
```

```python
severe_delays = df[df["ARR_DELAY"] > 60]

print("Flights delayed more than 60 minutes:", len(severe_delays))
```

```python
print("===== AIRLINE ANALYTICS SUMMARY =====")

print("Total Flights:", len(df))
print("Total Columns:", len(df.columns))
print("Average Departure Delay:", round(df["DEP_DELAY"].mean(), 2))
print("Average Arrival Delay:", round(df["ARR_DELAY"].mean(), 2))
print("Cancelled Flights:", df["CANCELLED"].sum())
print("Diverted Flights:", df["DIVERTED"].sum())
print("Delayed Flights:", len(delayed_flights))
print("Severely Delayed Flights:", len(severe_delays))
```

```python
airline_delay = df.groupby("AIRLINE")[["DEP_DELAY", "ARR_DELAY"]].mean().sort_values(
    "ARR_DELAY", ascending=False
)

airline_delay.plot(kind="bar", figsize=(12,6))

plt.title("Average Departure and Arrival Delay by Airline")
plt.xlabel("Airline")
plt.ylabel("Average Delay (minutes)")
plt.xticks(rotation=45)
plt.legend(["Departure Delay", "Arrival Delay"])
plt.tight_layout()
plt.show()
```

```python
route_counts = (
    df.groupby(["ORIGIN", "DEST"])
      .size()
      .sort_values(ascending=False)
      .head(10)
)

route_counts.index = [
    f"{origin} → {dest}"
    for origin, dest in route_counts.index
]

route_counts.plot(kind="bar", figsize=(12, 6))

plt.title("Top 10 Busiest Flight Routes")
plt.xlabel("Route")
plt.ylabel("Number of Flights")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
delay_cols = [
    "DELAY_DUE_CARRIER",
    "DELAY_DUE_WEATHER",
    "DELAY_DUE_NAS",
    "DELAY_DUE_SECURITY",
    "DELAY_DUE_LATE_AIRCRAFT"
]

delay_data = df[delay_cols].sum().sort_values(ascending=False)

delay_data.plot(kind="bar", figsize=(10, 6))

plt.title("Total Delay by Cause")
plt.xlabel("Delay Cause")
plt.ylabel("Total Delay (minutes)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
flight_status = df["DEP_DELAY"].apply(
    lambda x: "On Time" if x <= 0 else "Delayed"
)

status_counts = flight_status.value_counts()

status_counts.plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7)
)

plt.title("On-Time vs Delayed Flights")
plt.ylabel("")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["DISTANCE"],
    df["ARR_DELAY"],
    alpha=0.3
)

plt.title("Flight Distance vs Arrival Delay")
plt.xlabel("Distance")
plt.ylabel("Arrival Delay (minutes)")
plt.tight_layout()
plt.show()
```

```python
monthly_data = df.groupby(
    df["FL_DATE"].dt.month if pd.api.types.is_datetime64_any_dtype(df["FL_DATE"])
    else pd.to_datetime(df["FL_DATE"]).dt.month
).agg(
    total_flights=("FL_DATE", "count"),
    avg_departure_delay=("DEP_DELAY", "mean"),
    avg_arrival_delay=("ARR_DELAY", "mean")
)

monthly_data.index.name = "Month"

monthly_data["avg_departure_delay"].plot(
    kind="line",
    marker="o",
    figsize=(10, 6)
)

plt.title("Monthly Average Departure Delay")
plt.xlabel("Month")
plt.ylabel("Average Delay (minutes)")
plt.xticks(monthly_data.index)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
taxi_data = df.groupby("AIRLINE")[[
    "TAXI_OUT",
    "TAXI_IN"
]].mean().sort_values("TAXI_OUT", ascending=False)

taxi_data.plot(
    kind="bar",
    figsize=(12, 6)
)

plt.title("Average Taxi Time by Airline")
plt.xlabel("Airline")
plt.ylabel("Average Taxi Time (minutes)")
plt.xticks(rotation=45)
plt.legend(["Taxi Out", "Taxi In"])
plt.tight_layout()
plt.show()
```
